## Generación sin Imágen Base

```python
from diffusers import StableDiffusionPipeline
import torch
from PIL import Image

pipe_normal = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
).to("cuda")

promtp_1 = pipe_normal(
    "A cyberpunk city skyline at night"
).images[0]

promtp_1.save("city_prompt.png")

promtp_2 = pipe_normal(
    "Mistic Temple"
).images[0]

promtp_2.save("mistic_prompt.png")

promtp_3 = pipe_normal(
    "A samurai paying respect in the temple"
).images[0]

promtp_3.save("samurai_prompt.png")
```

## Canny

```python
from diffusers import ControlNetModel, StableDiffusionPipeline, StableDiffusionControlNetPipeline
import torch
from PIL import Image
from controlnet_aux import CannyDetector

controlnet = ControlNetModel.from_pretrained(
    "lllyasviel/sd-controlnet-canny",
    torch_dtype=torch.float16
)

image = Image.open("/content/Canny_Base.jpg")
detector = CannyDetector()
condition_image = detector(image)
condition_image.save("canny.png")

pipe = StableDiffusionControlNetPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    controlnet=controlnet,
    torch_dtype=torch.float16
).to("cuda")

result = pipe(
    prompt="A cyberpunk city skyline at night",
    image=condition_image,
    num_inference_steps=30
).images[0]

result.save("canny_result.png")
```

## Depth

```python
from diffusers import ControlNetModel, StableDiffusionControlNetPipeline
import torch
from PIL import Image
from controlnet_aux import MidasDetector

controlnet = ControlNetModel.from_pretrained(
    "lllyasviel/sd-controlnet-depth",
    torch_dtype=torch.float16
)

pipe = StableDiffusionControlNetPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    controlnet=controlnet,
    torch_dtype=torch.float16
).to("cuda")

midas = MidasDetector.from_pretrained("lllyasviel/Annotators")
midas.to("cuda")

image = Image.open("/content/Depth_Base.jpg")
depth_image = midas(image)
depth_image.save("depth.png")

result = pipe(
    prompt="Mistic Temple",
    image=depth_image,
    num_inference_steps=30
).images[0]

result.save("depth_result.png")
```
